```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("training_set3.csv", index_col=0)

features = ["ax", "ay", "az", "gx", "gy", "gz"]

class_stats = df.groupby("label")[["ax", "ay", "az", "gx", "gy", "gz"]].agg(
    ["mean", "std", "min", "max"]
)

#Maybe not need
id_stats = df.groupby(["label", "recording_id"]).agg({
    "ax": ["mean", "std", "min", "max"],
    "ay": ["mean", "std", "min", "max"],
    "az": ["mean", "std", "min", "max"],
    "gx": ["mean", "std", "min", "max"],
    "gy": ["mean", "std", "min", "max"],
    "gz": ["mean", "std", "min", "max"]
    }
)

def plot_distribution_by_class(df, column):
    classes = df["label"].unique()

    for classname in classes:
        data = df[df["label"] == classname]

        plt.hist(
            data[column],
            bins=30,
            alpha=0.5,
            label=classname
        )

    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"{column} distribution by class")
    plt.legend()
    plt.show()


class_stats
#id_stats
```

```python
#*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=* 
# 
for col in features:
    recording_mean = (
        df.groupby(["label", "recording_id"])[col]
          .mean()
          .reset_index()
    )

    #Box plot: Where the means of recs cluster
    recording_mean.boxplot(column=col, by="label")
    
    plt.title(f"Mean {col} per recording")
    plt.suptitle("")
    plt.xlabel("Class")
    plt.ylabel(f"Mean {col}")
    plt.show()

    #Scatter plot: Variation between means of recordings
    sns.stripplot(
        data=recording_mean,
        x="label",
        y=col
    )
    
    plt.xlabel("Class")
    plt.ylabel(f"Mean {col} ")
    plt.title(f"Mean {col} per recording")
    plt.show()
```

```python
#*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=* 
# Time series plot to see if recordings are consistent
#*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=*=* 

for recording_id, recording in df.groupby("recording_id"):
    col = "ay"
    plt.plot(recording["time"], recording[col])
    plt.xlabel("Time")
    plt.ylabel(f"{col}")
    plt.title(f"Recording {recording_id}")
    plt.show()
```
