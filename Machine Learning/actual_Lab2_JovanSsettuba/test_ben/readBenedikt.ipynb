```python
import os
import glob
import pandas as pd


def add_labels(folder, label, rec_id):
    files = glob.glob(os.path.join(folder, "*.csv"))
    dataframes = []

    for file in files:
        df = pd.read_csv(file)

        df["label"] = label
        df["recording_id"] = rec_id
        dataframes.append(df)
        
    return pd.concat(dataframes, ignore_index=False)

root_folder = r"C:\Users\JOSS0001\OneDrive - Högskolan Kristianstad\CTerminator 2\actual_Lab2_JovanSsettuba\test_ben"

all_data = []

for root, folders, files in os.walk(root_folder):

    csv_files = glob.glob(os.path.join(root, "*.csv"))

    if not csv_files:
        continue

    # Class name
    label = os.path.basename(root)

    for file in csv_files:

        # Example:
        # walking11 2026.csv
        filename = os.path.basename(file)

        # Get recording ID
        recording_id = filename.split(" ")[0]

        # Read CSV
        df = pd.read_csv(file)

        # Add label and recording ID
        df["label"] = label
        df["recording_id"] = recording_id

        # Add to list
        all_data.append(df)


# Combine all recordings
final_df = pd.concat(all_data, ignore_index=True)

print(final_df.head())
print(final_df.columns)

final_df.to_csv("C:\\Users\\JOSS0001\\OneDrive - Högskolan Kristianstad\\CTerminator 2\\actual_Lab2_JovanSsettuba\\ben_test_set.csv") #Save a frame to csv
```
