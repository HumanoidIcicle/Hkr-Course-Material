```python
# Split the data into training and test sets
import numpy as np
import pandas as pd

np.random.seed(42)
df = pd.read_csv("neuroTraining.csv", index_col=0)

val_ids = []
train_ids = []

for label in df["label"].unique(): #Gets all labels in df
    ids = df[df["label"] == label]["recording_id"].unique() #All recs for each label

    #Pick 2 recs for validation set
    val_selected = np.random.choice(ids, size=2, replace=False)
    val_ids.extend(val_selected)

    #Remove val recs
    for i in ids:
        if i not in val_selected:
            train_ids.append(i)

training_df = df[df["recording_id"].isin(train_ids)]
val_df = df[df["recording_id"].isin(val_ids)]

#Save the separate sets
training_df.to_csv("ntrain_set.csv")
val_df.to_csv("nval_set.csv")
```
