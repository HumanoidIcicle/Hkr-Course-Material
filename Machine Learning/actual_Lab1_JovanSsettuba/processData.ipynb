```python
import os
import glob
import pandas as pd

def combineFiles(folder, className):
    files = glob.glob(os.path.join(folder, "*.csv")) # find all csv files in a folder
    
    dataframes = [] # Read each csv

    for file in files:
        df = pd.read_csv(file, index_col=0)
        
        df.columns = ["time", "ax", "ay", "az", "gx", "gy", "gz"]
        df["label"] = className #Add class label column

        dataframes.append(df)

    finalFrame = pd.concat(dataframes, ignore_index=True) #Combine all the dataframes

    #df.to_csv(fileProcessed) #Save a frame to csv
    
    return finalFrame
```

```python
folder = "C:\\Users\\JOSS0001\\OneDrive - Högskolan Kristianstad\\CTerminator 2\\actual_Lab1\\circle"
className = "circle"
x = combineFiles(folder, className)

folder = "C:\\Users\\JOSS0001\\OneDrive - Högskolan Kristianstad\\CTerminator 2\\actual_Lab1\\table"
className = "table"
y = combineFiles(folder, className)

folder = "C:\\Users\\JOSS0001\\OneDrive - Högskolan Kristianstad\\CTerminator 2\\actual_Lab1\\upright"
className = "upright"
z = combineFiles(folder, className)
```
