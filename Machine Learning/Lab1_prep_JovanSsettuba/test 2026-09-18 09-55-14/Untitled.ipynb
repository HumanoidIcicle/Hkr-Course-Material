```python
import pandas as pd
import matplotlib.pyplot as plt

df2 = pd.read_csv("Accelerometer.csv")

plt.plot(df2["Acceleration x (m/s^2)"], label="X")
plt.plot(df2["Acceleration y (m/s^2)"], label="Y")
plt.plot(df2["Acceleration z (m/s^2)"], label="Z")

plt.xlabel("Sample")
plt.ylabel("Acceleration")
plt.legend()
#df2
```
